```python
"""
==============================================================
Nombre del archivo: red_neuronal_and.py
Autor: Orlando Munguia García.
Fecha de creación: 2025-05-01
Descripción:
    Este script implementa una red neuronal simple (perceptrón) que aprende
    la función lógica AND mediante el algoritmo de backpropagation.

    Esta red neuronal:
    - Tiene dos entradas binarias.
    - Usa una sola neurona con función de activación sigmoide.
    - Se entrena con descenso de gradiente y backpropagation.
    - Predice valores cercanos a 0 o 1, como salida de la función AND.

Requisitos:
    - Python 3.x
    - NumPy (para operaciones vectorizadas)

==============================================================
"""

import numpy as np  # Librería fundamental para cálculos numéricos en matrices y vectores

# ----------------------------------------
# Funciones auxiliares
# ----------------------------------------

def sigmoid(x):
    """
    Función de activación sigmoide.
    Transforma cualquier valor real a un rango entre 0 y 1.
    Es útil para problemas de clasificación binaria.
    """
    return 1 / (1 + np.exp(-x))

def sigmoid_derivada(x):
    """
    Derivada de la función sigmoide.
    Se utiliza durante el backpropagation para ajustar los pesos
    en función del gradiente del error.
    """
    return x * (1 - x)

# ----------------------------------------
# Datos de entrada y salida esperada
# ----------------------------------------

# Entradas de la función AND
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

# Salidas correspondientes para la función lógica AND
# Solo da 1 si ambas entradas son 1
y = np.array([[0],
              [0],
              [0],
              [1]])

# ----------------------------------------
# Inicialización de parámetros
# ----------------------------------------

np.random.seed(42)  # Fijamos la semilla para reproducir resultados

# Pesos iniciales aleatorios para las dos entradas (1 neurona)
pesos = np.random.uniform(-1, 1, (2, 1))

# Bias (umbral de activación), también inicializado aleatoriamente
bias = np.random.uniform(-1, 1)

# Tasa de aprendizaje: define qué tan grande es cada ajuste en pesos
alpha = 0.1

# ----------------------------------------
# Entrenamiento del modelo
# ----------------------------------------

# Entrenamos por 10,000 épocas para que la red aprenda la función AND
for epoch in range(10000):
    # ---- Paso 1: Propagación hacia adelante (Forward Pass) ----
    z = np.dot(X, pesos) + bias    # Calcula la entrada neta a la neurona
    salida = sigmoid(z)            # Aplica función de activación sigmoide

    # ---- Paso 2: Cálculo del error ----
    error = y - salida             # Diferencia entre salida deseada y actual

    # ---- Paso 3: Retropropagación del error (Backpropagation) ----
    # Se calcula la corrección en la salida en función del error y la derivada
    d_salida = error * sigmoid_derivada(salida)

    # ---- Paso 4: Actualización de pesos y bias ----
    pesos += np.dot(X.T, d_salida) * alpha    # Ajuste de pesos con regla delta
    bias += np.sum(d_salida) * alpha          # Ajuste del bias con la suma de los errores

# ----------------------------------------
# Evaluación del modelo entrenado
# ----------------------------------------

print("Pesos entrenados:", pesos.ravel())        # Muestra los pesos finales ajustados
print("Bias entrenado:", bias)                   # Muestra el bias aprendido
print("Predicciones:")
# Calcula y muestra las salidas con los parámetros entrenados, redondeadas a 3 decimales
print(np.round(sigmoid(np.dot(X, pesos) + bias), 3))
```
